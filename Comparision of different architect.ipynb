```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

def explain_alexnet():
    print("=== AlexNet (2012) ===")
    print("First CNN to win ImageNet, 8 layers (5 conv, 3 dense)")
    
    # Simplified AlexNet architecture
    alexnet = nn.Sequential(
        # Conv1
        nn.Conv2d(3, 96, kernel_size=11, stride=4),
        nn.ReLU(),
        nn.MaxPool2d(kernel_size=3, stride=2),
        # Conv2
        nn.Conv2d(96, 256, kernel_size=5, padding=2),
        nn.ReLU(),
        nn.MaxPool2d(kernel_size=3, stride=2),
        # Conv3-5
        nn.Conv2d(256, 384, kernel_size=3, padding=1),
        nn.ReLU(),
        nn.Conv2d(384, 384, kernel_size=3, padding=1),
        nn.ReLU(),
        nn.Conv2d(384, 256, kernel_size=3, padding=1),
        nn.ReLU(),
        nn.MaxPool2d(kernel_size=3, stride=2),
        # Flatten
        nn.Flatten(),
        # Dense layers
        nn.Linear(256*6*6, 4096),
        nn.ReLU(),
        nn.Dropout(0.5),
        nn.Linear(4096, 4096),
        nn.ReLU(),
        nn.Dropout(0.5),
        nn.Linear(4096, 1000)  # ImageNet classes
    )
    
    # Visualize architecture
    print("\nArchitecture Visualization:")
    input_tensor = torch.randn(1, 3, 227, 227)  # AlexNet input size
    for layer in alexnet:
        input_tensor = layer(input_tensor)
        print(f"{layer.__class__.__name__}: Output shape {tuple(input_tensor.shape)}")
    
    # Key features
    print("\nKey Features:")
    print("- First use of ReLU in deep CNN")
    print("- Used dropout for regularization")
    print("- Trained on 2 GPUs (split architecture)")

# explain_alexnet()

def explain_imagenet():
    print("\n=== ImageNet ===")
    print("Large visual database (14M+ images, 20k+ categories)")
    
    # Show sample categories
    categories = [
        "n01440764: tench (fish)", "n02102040: English springer (dog)",
        "n02979186: cassette player", "n03000684: chainsaw",
        "n03417042: garbage truck", "n03938244: pillow"
    ]
    
    print("\nSample Categories:")
    for cat in categories:
        print(f"- {cat}")
    
    # ImageNet stats
    print("\nKey Facts:")
    print("- 1000 classes in competition")
    print("- Images resized to 256x256")
    print("- Used for benchmarking CNN architectures")
    
    # Visualization idea (would need actual images)
    print("\nTeaching Tip:")
    print("Show sample images from different categories to demonstrate variety")

# explain_imagenet()

def explain_unet():
    print("\n=== U-Net (2015) ===")
    print("Originally for biomedical image segmentation, encoder-decoder with skip connections")
    
    # Simplified U-Net components
    class DoubleConv(nn.Module):
        def __init__(self, in_ch, out_ch):
            super().__init__()
            self.conv = nn.Sequential(
                nn.Conv2d(in_ch, out_ch, 3, padding=1),
                nn.ReLU(),
                nn.Conv2d(out_ch, out_ch, 3, padding=1),
                nn.ReLU()
            )
        def forward(self, x):
            return self.conv(x)
    
    print("\nKey Components:")
    print("- Contracting path (encoder): captures context")
    print("- Expanding path (decoder): enables precise localization")
    print("- Skip connections: combine high-res features from encoder with decoder")
    
    # Visualize data flow
    print("\nData Flow Example (simplified):")
    print("Input (1,572,572) -> Conv1 -> (64,568,568) -> MaxPool -> (64,284,284)")
    print("... (bottleneck) ... Upsample -> (128,280,280) -> Concatenate -> (256,280,280)")
    print("Final Output: (n_classes, 388,388)")
    
    # Teaching demo idea
    print("\nInteractive Demo Idea:")
    print("1. Show medical image (e.g., MRI scan)")
    print("2. Show ground truth segmentation")
    print("3. Visualize how U-Net combines features at different scales")

# explain_unet()
def explain_resnet():
    print("\n=== ResNet (2015) ===")
    print("Revolutionary 'skip connections' enabling very deep networks (150+ layers)")
    
    # Residual block implementation
    class ResidualBlock(nn.Module):
        def __init__(self, in_ch, out_ch, stride=1):
            super().__init__()
            self.conv1 = nn.Conv2d(in_ch, out_ch, kernel_size=3, stride=stride, padding=1)
            self.bn1 = nn.BatchNorm2d(out_ch)
            self.conv2 = nn.Conv2d(out_ch, out_ch, kernel_size=3, padding=1)
            self.bn2 = nn.BatchNorm2d(out_ch)
            
            self.shortcut = nn.Sequential()
            if stride != 1 or in_ch != out_ch:
                self.shortcut = nn.Sequential(
                    nn.Conv2d(in_ch, out_ch, kernel_size=1, stride=stride),
                    nn.BatchNorm2d(out_ch)
                )
        
        def forward(self, x):
            out = torch.relu(self.bn1(self.conv1(x)))
            out = self.bn2(self.conv2(out))
            out += self.shortcut(x)  # The skip connection!
            return torch.relu(out)
    
    print("\nResidual Block Code Breakdown:")
    print("- Two conv layers with batch norm")
    print("- Shortcut connection for original input")
    print("- Element-wise addition of shortcut to conv output")
    
    # Why it works
    print("\nWhy ResNet Works:")
    print("- Solves vanishing gradient problem in deep networks")
    print("- Networks can learn 'identity functions' easily")
    print("- Enables gradient flow through shortcuts")
    
    # Teaching visualization
    print("\nTeaching Visualization:")
    print("Compare training of:")
    print("1. Plain 34-layer network (poor performance)")
    print("2. ResNet-34 (much better)")

# explain_resnet()
```
